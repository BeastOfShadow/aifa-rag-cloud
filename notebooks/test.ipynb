```python
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)
```

```python
# 3 CSV in 3 variabili. Separatore ';'. dtype=str -> non perde zeri iniziali nei codici.
BASE = "../datasets"

atc  = pd.read_csv(f"{BASE}/atc.csv",                  sep=";", dtype=str)
pa   = pd.read_csv(f"{BASE}/PA_confezioni.csv",        sep=";", dtype=str)
conf = pd.read_csv(f"{BASE}/confezioni_fornitura.csv", sep=";", dtype=str)

print("atc ", atc.shape)
print("pa  ", pa.shape)
print("conf", conf.shape)
```

```python
def peek(name, df):
    print(f"===== {name}  shape={df.shape} =====")
    print("null per colonna:")
    print(df.isna().sum())
    print("\nvalori unici per colonna:")
    print(df.nunique())
    display(df.head())
```

## `atc` — classificazione ATC

Tabella di lookup: codice ATC &rarr; descrizione della categoria terapeutica.

| Colonna | Descrizione | Esempio |
|---|---|---|
| `CODICE_ATC` | Codice classificazione ATC (Anatomico Terapeutico Chimico). Livello variabile (categoria o principio) | `C02AA01` |
| `DESCRIZIONE` | Nome della categoria / sostanza | `RESCINNAMINA` |

```python
peek("atc", atc)
```

## `pa` — principi attivi per confezione

Ogni riga = un principio attivo associato a una confezione (identificata da codice AIC).

| Colonna | Descrizione | Esempio |
|---|---|---|
| `CODICE_AIC` | Codice AIC della confezione (Autorizzazione Immissione in Commercio) | `000152037` |
| `PRINCIPIO_ATTIVO` | Nome principio attivo (`N.D.` = non disponibile) | `ACIDO ACETILSALICILICO` |
| `QUANTITA` | Quantita del principio | `0.0` |
| `UNITA_MISURA` | Unita di misura della quantita | `N.D.` |

```python
peek("pa", pa)
```

## `conf` — confezioni e fornitura

Anagrafica delle confezioni di farmaci: denominazione, ditta, ATC, regime di fornitura, link ai fogli.

| Colonna | Descrizione | Esempio |
|---|---|---|
| `CODICE_AIC` | Codice AIC confezione (chiave verso `pa`) | `000367045` |
| `COD_FARMACO` | Codice farmaco | `000367` |
| `COD_CONFEZIONE` | Codice confezione | `045` |
| `DENOMINAZIONE` | Nome commerciale | `TISANA KELEMATA` |
| `DESCRIZIONE` | Descrizione confezione | `10 BUSTINE FILTRO G 2` |
| `CODICE_DITTA` | Codice azienda | `2934` |
| `RAGIONE_SOCIALE` | Nome azienda titolare | `KELEMATA S.R.L.` |
| `STATO_AMMINISTRATIVO` | Stato AIC | `Autorizzata` |
| `TIPO_PROCEDURA` | Procedura autorizzativa | `Procedura Nazionale` |
| `FORMA` | Forma farmaceutica | `Tisana` |
| `CODICE_ATC` | Codice ATC (chiave verso `atc`) | `A06AB06` |
| `PA_ASSOCIATI` | Principi attivi associati (testo) | `SENNA FOGLIA` |
| `FORNITURA` | Regime di fornitura | `Medicinali non soggetti...` |
| `LINK_FI` | Link Foglio Illustrativo | url |
| `LINK_RCP` | Link Riassunto Caratteristiche Prodotto | url |

```python
peek("conf", conf)
```
